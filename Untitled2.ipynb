```python
%%writefile modules/utils.py
"""
Utility functions for the dashboard
"""

import logging
import config
from datetime import datetime
from typing import Optional


def setup_logging():
    """Setup logging configuration"""
    logging.basicConfig(
        level=getattr(logging, config.LOG_LEVEL),
        format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
        handlers=[
            logging.FileHandler(config.LOG_FILE),
            logging.StreamHandler()
        ]
    )
    return logging.getLogger(__name__)


def format_number(num: Optional[float], decimals: int = 2) -> str:
    """Format number with commas and decimals"""
    if num is None:
        return "N/A"

    if num >= 1_000_000:
        return f"${num/1_000_000:.{decimals}f}M"
    elif num >= 1_000:
        return f"${num/1_000:.{decimals}f}K"
    else:
        return f"${num:.{decimals}f}"


def format_percentage(num: Optional[float], decimals: int = 2, show_sign: bool = True) -> str:
    """Format percentage with optional sign"""
    if num is None:
        return "N/A"

    sign = "+" if num > 0 and show_sign else ""
    return f"{sign}{num:.{decimals}f}%"


def get_color_for_change(change: float) -> str:
    """Get color based on positive/negative change"""
    if change > 0:
        return config.COLORS["positive"]
    elif change < 0:
        return config.COLORS["negative"]
    else:
        return config.COLORS["neutral"]


def calculate_percentage_change(old_value: float, new_value: float) -> float:
    """Calculate percentage change between two values"""
    if old_value == 0:
        return 0.0
    return ((new_value - old_value) / old_value) * 100


def timestamp_to_datetime(timestamp: int) -> datetime:
    """Convert unix timestamp to datetime"""
    return datetime.fromtimestamp(timestamp)


def datetime_to_timestamp(dt: datetime) -> int:
    """Convert datetime to unix timestamp"""
    return int(dt.timestamp())


def truncate_text(text: str, max_length: int = 50) -> str:
    """Truncate text with ellipsis"""
    if len(text) <= max_length:
        return text
    return text[:max_length-3] + "..."


def safe_divide(numerator: float, denominator: float, default: float = 0.0) -> float:
    """Safe division with default value"""
    try:
        if denominator == 0:
            return default
        return numerator / denominator
    except:
        return default
```

```python
%%writefile modules/api_client.py
# Copy the ENTIRE api_client.py code from my previous response here
# (I'll put the full code - it's long but necessary)

"""
Polymarket API Client with caching and retry logic
"""

import requests
import time
import logging
from typing import Dict, List, Optional, Any
from datetime import datetime, timedelta
import json
import config


class PolymarketAPI:
    """Handle all Polymarket API interactions"""

    def __init__(self):
        self.logger = logging.getLogger(__name__)
        self.cache = {}
        self.session = requests.Session()
        self.session.headers.update({
            'User-Agent': 'PolymarketDashboard/1.0'
        })

    def _make_request(self, url: str, params: Optional[Dict] = None,
                      retries: int = config.MAX_RETRIES) -> Optional[Dict]:
        """Make API request with retry logic"""

        cache_key = f"{url}_{json.dumps(params or {})}"
        if cache_key in self.cache:
            cached_data, cache_time = self.cache[cache_key]
            if (datetime.now() - cache_time).seconds < config.CACHE_DURATION_SECONDS:
                self.logger.debug(f"Cache hit for {url}")
                return cached_data

        for attempt in range(retries):
            try:
                self.logger.debug(f"API request to {url} (attempt {attempt + 1})")
                response = self.session.get(
                    url,
                    params=params,
                    timeout=config.API_TIMEOUT_SECONDS
                )
                response.raise_for_status()
                data = response.json()

                self.cache[cache_key] = (data, datetime.now())

                return data

            except requests.exceptions.RequestException as e:
                self.logger.warning(f"API request failed (attempt {attempt + 1}): {e}")

                if attempt < retries - 1:
                    delay = config.RETRY_DELAY_SECONDS * (config.RETRY_BACKOFF_MULTIPLIER ** attempt)
                    time.sleep(delay)
                else:
                    self.logger.error(f"All retry attempts failed for {url}")
                    return None

        return None

    def get_markets(self, limit: int = 100, active: bool = True) -> List[Dict]:
        """Fetch list of markets"""
        params = {
            "limit": min(limit, config.MAX_MARKETS_TO_TRACK),
            "active": str(active).lower(),
            "closed": "false"
        }

        data = self._make_request(config.ENDPOINTS["markets"], params)

        if data and isinstance(data, list):
            self.logger.info(f"Fetched {len(data)} markets")
            return data

        self.logger.warning("Failed to fetch markets or empty response")
        return []

    def get_market_by_id(self, market_id: str) -> Optional[Dict]:
        """Fetch specific market by ID"""
        url = f"{config.ENDPOINTS['markets']}/{market_id}"
        return self._make_request(url)

    def get_market_prices(self, token_ids: List[str]) -> Dict[str, float]:
        """Fetch current prices for multiple tokens"""
        if not token_ids:
            return {}

        params = {"token_id": ",".join(token_ids)}
        data = self._make_request(config.ENDPOINTS["prices"], params)

        if not data:
            return {}

        prices = {}
        try:
            if isinstance(data, dict):
                for token_id, price_data in data.items():
                    if isinstance(price_data, dict) and 'price' in price_data:
                        prices[token_id] = float(price_data['price'])
                    elif isinstance(price_data, (int, float, str)):
                        prices[token_id] = float(price_data)
            elif isinstance(data, list):
                for item in data:
                    if 'token_id' in item and 'price' in item:
                        prices[item['token_id']] = float(item['price'])
        except Exception as e:
            self.logger.error(f"Error parsing prices: {e}")

        return prices

    def get_price_history(self, token_id: str,
                          start_time: Optional[int] = None,
                          end_time: Optional[int] = None) -> List[Dict]:
        """Fetch historical price data for a token"""
        params = {"token_id": token_id}

        if start_time:
            params["start_time"] = start_time
        if end_time:
            params["end_time"] = end_time

        data = self._make_request(config.ENDPOINTS["prices_history"], params)

        if data and isinstance(data, list):
            return data

        return []

    def get_orderbook(self, token_id: str) -> Optional[Dict]:
        """Fetch orderbook for a token"""
        params = {"token_id": token_id}
        return self._make_request(config.ENDPOINTS["orderbook"], params)

    def clear_cache(self):
        """Clear API cache"""
        self.cache = {}
        self.logger.info("API cache cleared")

    def get_cache_stats(self) -> Dict[str, int]:
        """Get cache statistics"""
        return {
            "cached_requests": len(self.cache),
            "cache_size_bytes": len(str(self.cache))
        }
```

```python
%%writefile modules/storage_manager.py
"""
Local storage manager for historical snapshots
"""

import json
import logging
from pathlib import Path
from typing import Dict, List, Optional
from datetime import datetime, timedelta
import config


class StorageManager:
    """Manage local JSON storage for market snapshots"""

    def __init__(self):
        self.logger = logging.getLogger(__name__)
        self.snapshots_dir = config.SNAPSHOTS_DIR
        self.cache_dir = config.CACHE_DIR

    def save_snapshot(self, market_id: str, data: Dict):
        """Save market snapshot to local storage"""
        try:
            market_file = self.snapshots_dir / f"{market_id}.json"

            snapshots = []
            if market_file.exists():
                with open(market_file, 'r') as f:
                    snapshots = json.load(f)

            snapshot = {
                "timestamp": int(datetime.now().timestamp()),
                "data": data
            }
            snapshots.append(snapshot)

            if len(snapshots) > config.MAX_SNAPSHOTS_PER_MARKET:
                snapshots = snapshots[-config.MAX_SNAPSHOTS_PER_MARKET:]

            with open(market_file, 'w') as f:
                json.dump(snapshots, f, indent=2)

            self.logger.debug(f"Saved snapshot for market {market_id}")

        except Exception as e:
            self.logger.error(f"Failed to save snapshot for {market_id}: {e}")

    def get_snapshots(self, market_id: str,
                      limit: Optional[int] = None,
                      since: Optional[datetime] = None) -> List[Dict]:
        """Get historical snapshots for a market"""
        try:
            market_file = self.snapshots_dir / f"{market_id}.json"

            if not market_file.exists():
                return []

            with open(market_file, 'r') as f:
                snapshots = json.load(f)

            if since:
                since_timestamp = int(since.timestamp())
                snapshots = [s for s in snapshots if s["timestamp"] >= since_timestamp]

            if limit:
                snapshots = snapshots[-limit:]

            return snapshots

        except Exception as e:
            self.logger.error(f"Failed to load snapshots for {market_id}: {e}")
            return []

    def get_latest_snapshot(self, market_id: str) -> Optional[Dict]:
        """Get most recent snapshot for a market"""
        snapshots = self.get_snapshots(market_id, limit=1)
        return snapshots[0] if snapshots else None

    def cleanup_old_snapshots(self):
        """Remove snapshots older than configured days"""
        try:
            cutoff_date = datetime.now() - timedelta(days=config.SNAPSHOT_CLEANUP_DAYS)
            cutoff_timestamp = int(cutoff_date.timestamp())

            cleaned_count = 0

            for market_file in self.snapshots_dir.glob("*.json"):
                with open(market_file, 'r') as f:
                    snapshots = json.load(f)

                original_count = len(snapshots)
                snapshots = [s for s in snapshots if s["timestamp"] >= cutoff_timestamp]

                if len(snapshots) < original_count:
                    with open(market_file, 'w') as f:
                        json.dump(snapshots, f, indent=2)
                    cleaned_count += original_count - len(snapshots)

            if cleaned_count > 0:
                self.logger.info(f"Cleaned {cleaned_count} old snapshots")

        except Exception as e:
            self.logger.error(f"Failed to cleanup old snapshots: {e}")

    def save_cache(self, key: str, data: any):
        """Save data to cache"""
        try:
            cache_file = self.cache_dir / f"{key}.json"
            with open(cache_file, 'w') as f:
                json.dump({
                    "timestamp": int(datetime.now().timestamp()),
                    "data": data
                }, f, indent=2)
        except Exception as e:
            self.logger.error(f"Failed to save cache {key}: {e}")

    def load_cache(self, key: str, max_age_seconds: int = 300) -> Optional[any]:
        """Load data from cache if not expired"""
        try:
            cache_file = self.cache_dir / f"{key}.json"

            if not cache_file.exists():
                return None

            with open(cache_file, 'r') as f:
                cached = json.load(f)

            age = datetime.now().timestamp() - cached["timestamp"]
            if age > max_age_seconds:
                return None

            return cached["data"]

        except Exception as e:
            self.logger.error(f"Failed to load cache {key}: {e}")
            return None

    def get_all_tracked_markets(self) -> List[str]:
        """Get list of all market IDs with snapshots"""
        return [f.stem for f in self.snapshots_dir.glob("*.json")]

    def get_storage_stats(self) -> Dict:
        """Get storage statistics"""
        try:
            snapshot_files = list(self.snapshots_dir.glob("*.json"))
            total_snapshots = 0

            for file in snapshot_files:
                with open(file, 'r') as f:
                    total_snapshots += len(json.load(f))

            return {
                "markets_tracked": len(snapshot_files),
                "total_snapshots": total_snapshots,
                "storage_mb": sum(f.stat().st_size for f in snapshot_files) / (1024 * 1024)
            }
        except Exception as e:
            self.logger.error(f"Failed to get storage stats: {e}")
            return {}
```

```python
%%writefile modules/data_processor.py
"""
Data processor for market data analysis
"""

import pandas as pd
import numpy as np
import logging
from typing import Dict, List, Optional
from datetime import datetime, timedelta
import config
from .utils import calculate_percentage_change, safe_divide


class DataProcessor:
    """Process and analyze market data"""

    def __init__(self):
        self.logger = logging.getLogger(__name__)

    def process_markets(self, markets_data: List[Dict]) -> pd.DataFrame:
        """Convert raw market data to DataFrame with calculated fields"""
        if not markets_data:
            return pd.DataFrame()

        try:
            processed = []

            for market in markets_data:
                tokens = market.get('tokens', [])

                yes_token = None
                no_token = None

                for token in tokens:
                    outcome = token.get('outcome', '').upper()
                    if outcome == 'YES':
                        yes_token = token
                    elif outcome == 'NO':
                        no_token = token

                if not yes_token or not no_token:
                    continue

                processed_market = {
                    'market_id': market.get('id', ''),
                    'question': market.get('question', 'Unknown'),
                    'yes_price': float(yes_token.get('price', 0)),
                    'no_price': float(no_token.get('price', 0)),
                    'yes_token_id': yes_token.get('token_id', ''),
                    'no_token_id': no_token.get('token_id', ''),
                    'volume': float(market.get('volume', 0)),
                    'liquidity': float(market.get('liquidity', 0)),
                    'active': market.get('active', True),
                    'end_date': market.get('end_date_iso', None),
                    'category': market.get('category', 'Unknown'),
                    'description': market.get('description', ''),
                }

                processed_market['probability'] = processed_market['yes_price'] * 100

                processed.append(processed_market)

            df = pd.DataFrame(processed)

            if not df.empty:
                df['volume_formatted'] = df['volume'].apply(lambda x: f"${x:,.0f}")
                df['liquidity_formatted'] = df['liquidity'].apply(lambda x: f"${x:,.0f}")

                self.logger.info(f"Processed {len(df)} markets")

            return df

        except Exception as e:
            self.logger.error(f"Error processing markets: {e}")
            return pd.DataFrame()
```

```python
%%writefile -a modules/data_processor.py

    def calculate_velocity(self, snapshots: List[Dict], window_seconds: int) -> float:
        """Calculate probability velocity (% change per minute)"""
        if len(snapshots) < 2:
            return 0.0

        try:
            now = datetime.now().timestamp()
            cutoff = now - window_seconds

            recent_snapshots = [
                s for s in snapshots
                if s['timestamp'] >= cutoff
            ]

            if len(recent_snapshots) < 2:
                return 0.0

            earliest = recent_snapshots[0]['data']
            latest = recent_snapshots[-1]['data']

            earliest_prob = earliest.get('probability', 0)
            latest_prob = latest.get('probability', 0)

            time_diff_minutes = (latest_prob - earliest_prob) / (window_seconds / 60)

            change = latest_prob - earliest_prob

            return safe_divide(change, time_diff_minutes, 0.0)

        except Exception as e:
            self.logger.error(f"Error calculating velocity: {e}")
            return 0.0

    def calculate_acceleration(self, snapshots: List[Dict]) -> float:
        """Calculate probability acceleration (change in velocity)"""
        if len(snapshots) < 3:
            return 0.0

        try:
            mid_point = len(snapshots) // 2

            early_snapshots = snapshots[:mid_point]
            late_snapshots = snapshots[mid_point:]

            early_velocity = self._get_velocity_from_snapshots(early_snapshots)
            late_velocity = self._get_velocity_from_snapshots(late_snapshots)

            return late_velocity - early_velocity

        except Exception as e:
            self.logger.error(f"Error calculating acceleration: {e}")
            return 0.0

    def _get_velocity_from_snapshots(self, snapshots: List[Dict]) -> float:
        """Helper to calculate velocity from snapshot list"""
        if len(snapshots) < 2:
            return 0.0

        first = snapshots[0]
        last = snapshots[-1]

        prob_change = last['data'].get('probability', 0) - first['data'].get('probability', 0)
        time_diff = (last['timestamp'] - first['timestamp']) / 60

        return safe_divide(prob_change, time_diff, 0.0)

    def calculate_volatility(self, snapshots: List[Dict], window_seconds: int = 3600) -> float:
        """Calculate probability volatility (standard deviation)"""
        if len(snapshots) < 2:
            return 0.0

        try:
            now = datetime.now().timestamp()
            cutoff = now - window_seconds

            recent_snapshots = [
                s for s in snapshots
                if s['timestamp'] >= cutoff
            ]

            if len(recent_snapshots) < 2:
                return 0.0

            probabilities = [s['data'].get('probability', 0) for s in recent_snapshots]
            return float(np.std(probabilities))

        except Exception as e:
            self.logger.error(f"Error calculating volatility: {e}")
            return 0.0

    def detect_consolidation(self, snapshots: List[Dict],
                            min_period: int = config.CONSOLIDATION_MIN_PERIOD) -> Dict:
        """Detect if market is in consolidation phase"""
        if len(snapshots) < 2:
            return {"is_consolidating": False, "range": 0, "duration": 0}

        try:
            now = datetime.now().timestamp()
            cutoff = now - min_period

            recent_snapshots = [
                s for s in snapshots
                if s['timestamp'] >= cutoff
            ]

            if len(recent_snapshots) < 2:
                return {"is_consolidating": False, "range": 0, "duration": 0}

            probabilities = [s['data'].get('probability', 0) for s in recent_snapshots]

            prob_range = max(probabilities) - min(probabilities)
            duration = recent_snapshots[-1]['timestamp'] - recent_snapshots[0]['timestamp']

            is_consolidating = prob_range <= config.CONSOLIDATION_THRESHOLD

            return {
                "is_consolidating": is_consolidating,
                "range": prob_range,
                "duration": duration,
                "min_prob": min(probabilities),
                "max_prob": max(probabilities)
            }

        except Exception as e:
            self.logger.error(f"Error detecting consolidation: {e}")
            return {"is_consolidating": False, "range": 0, "duration": 0}

    def calculate_trend_consistency(self, snapshots: List[Dict]) -> float:
        """Calculate how consistent the trend direction is (0-1)"""
        if len(snapshots) < 3:
            return 0.0

        try:
            changes = []
            for i in range(1, len(snapshots)):
                prev_prob = snapshots[i-1]['data'].get('probability', 0)
                curr_prob = snapshots[i]['data'].get('probability', 0)
                changes.append(curr_prob - prev_prob)

            if not changes:
                return 0.0

            positive_changes = sum(1 for c in changes if c > 0)
            negative_changes = sum(1 for c in changes if c < 0)

            total_changes = len(changes)
            consistency = max(positive_changes, negative_changes) / total_changes

            return consistency

        except Exception as e:
            self.logger.error(f"Error calculating consistency: {e}")
            return 0.0
```

```python
%%writefile modules/signal_engine.py
"""
Signal detection engine for identifying rapid repricing events
"""

import logging
from typing import Dict, List, Optional
from datetime import datetime, timedelta
import config
from .data_processor import DataProcessor


class SignalEngine:
    """Detect market movement signals and generate alerts"""

    def __init__(self, data_processor: DataProcessor):
        self.logger = logging.getLogger(__name__)
        self.processor = data_processor

    def analyze_market(self, market_data: Dict, snapshots: List[Dict]) -> Dict:
        """
        Comprehensive market analysis with signal detection
        Returns signal data and alerts
        """
        if len(snapshots) < 2:
            return self._empty_analysis()

        try:
            analysis = {
                "market_id": market_data.get('market_id', ''),
                "question": market_data.get('question', ''),
                "current_probability": market_data.get('probability', 0),
                "timestamp": int(datetime.now().timestamp()),
            }

            # Calculate velocities for different time windows
            velocities = {}
            for window_name, window_seconds in config.VELOCITY_WINDOWS.items():
                velocities[window_name] = self.processor.calculate_velocity(
                    snapshots, window_seconds
                )
            analysis["velocities"] = velocities

            # Calculate acceleration
            analysis["acceleration"] = self.processor.calculate_acceleration(snapshots)

            # Calculate volatility
            analysis["volatility"] = self.processor.calculate_volatility(snapshots)

            # Detect consolidation
            consolidation = self.processor.detect_consolidation(snapshots)
            analysis["consolidation"] = consolidation

            # Calculate trend consistency
            analysis["consistency"] = self.processor.calculate_trend_consistency(snapshots)

            # Detect breakout
            analysis["breakout"] = self._detect_breakout(
                market_data, snapshots, consolidation
            )

            # Calculate signal score
            analysis["signal_score"] = self._calculate_signal_score(analysis, market_data)

            # Generate alerts
            analysis["alerts"] = self._generate_alerts(analysis)

            # Determine signal strength
            analysis["signal_strength"] = self._get_signal_strength(analysis["signal_score"])

            return analysis

        except Exception as e:
            self.logger.error(f"Error analyzing market: {e}")
            return self._empty_analysis()

    def _detect_breakout(self, market_data: Dict, snapshots: List[Dict],
                        consolidation: Dict) -> Dict:
        """Detect if market is breaking out from consolidation"""
        if len(snapshots) < 2:
            return {"is_breakout": False, "direction": None, "magnitude": 0}

        try:
            # Check if was recently consolidating
            was_consolidating = consolidation.get("is_consolidating", False)

            # Get recent price movement
            recent_velocity = self.processor.calculate_velocity(
                snapshots, config.VELOCITY_WINDOWS["5m"]
            )

            current_prob = market_data.get('probability', 0)

            # Breakout criteria:
            # 1. Was consolidating
            # 2. Strong velocity
            # 3. Moved outside consolidation range

            is_breakout = False
            direction = None
            magnitude = 0

            if was_consolidating and abs(recent_velocity) > config.BREAKOUT_THRESHOLD:
                consolidation_max = consolidation.get("max_prob", current_prob)
                consolidation_min = consolidation.get("min_prob", current_prob)

                if current_prob > consolidation_max:
                    is_breakout = True
                    direction = "UP"
                    magnitude = current_prob - consolidation_max
                elif current_prob < consolidation_min:
                    is_breakout = True
                    direction = "DOWN"
                    magnitude = consolidation_min - current_prob

            return {
                "is_breakout": is_breakout,
                "direction": direction,
                "magnitude": magnitude,
                "velocity": recent_velocity
            }

        except Exception as e:
            self.logger.error(f"Error detecting breakout: {e}")
            return {"is_breakout": False, "direction": None, "magnitude": 0}

    def _calculate_signal_score(self, analysis: Dict, market_data: Dict) -> float:
        """
        Calculate overall signal score (0-100)
        Higher score = stronger signal for rapid repricing
        """
        try:
            score = 0.0

            # Velocity component (0-30 points)
            velocity_1m = abs(analysis["velocities"].get("1m", 0))
            velocity_score = min(velocity_1m * 3, 30)  # 10% per minute = 30 points
            score += velocity_score * config.SIGNAL_WEIGHTS["velocity"]

            # Acceleration component (0-25 points)
            acceleration = abs(analysis.get("acceleration", 0))
            acceleration_score = min(acceleration * 5, 25)
            score += acceleration_score * config.SIGNAL_WEIGHTS["acceleration"]

            # Volume component (0-20 points)
            volume = market_data.get('volume', 0)
            volume_score = min(volume / 10000, 20)  # $100k = 20 points
            score += volume_score * config.SIGNAL_WEIGHTS["volume"]

            # Liquidity component (0-15 points)
            liquidity = market_data.get('liquidity', 0)
            liquidity_score = min(liquidity / 10000, 15)
            score += liquidity_score * config.SIGNAL_WEIGHTS["liquidity"]

            # Consistency component (0-10 points)
            consistency = analysis.get("consistency", 0)
            consistency_score = consistency * 10
            score += consistency_score * config.SIGNAL_WEIGHTS["consistency"]

            # Breakout bonus
            if analysis.get("breakout", {}).get("is_breakout"):
                score += 20

            return min(score, 100.0)

        except Exception as e:
            self.logger.error(f"Error calculating signal score: {e}")
            return 0.0

    def _generate_alerts(self, analysis: Dict) -> List[str]:
        """Generate alert messages based on analysis"""
        alerts = []

        try:
            # Velocity alerts
            velocity_1m = analysis["velocities"].get("1m", 0)
            if abs(velocity_1m) >= config.VELOCITY_ALERT_THRESHOLDS["high"]:
                alerts.append("🚨 EXTREME VELOCITY DETECTED")
            elif abs(velocity_1m) >= config.VELOCITY_ALERT_THRESHOLDS["medium"]:
                alerts.append("⚠️ High Velocity Movement")
            elif abs(velocity_1m) >= config.VELOCITY_ALERT_THRESHOLDS["low"]:
                alerts.append("📈 Elevated Velocity")

            # Acceleration alerts
            acceleration = analysis.get("acceleration", 0)
            if abs(acceleration) > 5:
                alerts.append("🔥 MOMENTUM EXPANDING")
            elif abs(acceleration) > 2:
                alerts.append("📊 Acceleration Detected")

            # Breakout alerts
            breakout = analysis.get("breakout", {})
            if breakout.get("is_breakout"):
                direction = breakout.get("direction", "")
                alerts.append(f"💥 BREAKOUT {direction}")

            # Consolidation breaking
            if analysis.get("consolidation", {}).get("is_consolidating"):
                if abs(velocity_1m) > 3:
                    alerts.append("⚡ POTENTIAL REPRICING STARTED")

            # High signal score
            signal_score = analysis.get("signal_score", 0)
            if signal_score >= 80:
                alerts.append("🎯 VERY STRONG SIGNAL")
            elif signal_score >= 60:
                alerts.append("✅ Strong Signal")

            # Volatility expansion
            volatility = analysis.get("volatility", 0)
            if volatility > 5:
                alerts.append("📉 High Volatility")

        except Exception as e:
            self.logger.error(f"Error generating alerts: {e}")

        return alerts

    def _get_signal_strength(self, score: float) -> str:
        """Get human-readable signal strength"""
        if score >= 80:
            return "VERY STRONG"
        elif score >= 60:
            return "STRONG"
        elif score >= 40:
            return "MODERATE"
        elif score >= 20:
            return "WEAK"
        else:
            return "NONE"

    def _empty_analysis(self) -> Dict:
        """Return empty analysis structure"""
        return {
            "market_id": "",
            "question": "",
            "current_probability": 0,
            "timestamp": int(datetime.now().timestamp()),
            "velocities": {},
            "acceleration": 0,
            "volatility": 0,
            "consolidation": {"is_consolidating": False},
            "breakout": {"is_breakout": False},
            "consistency": 0,
            "signal_score": 0,
            "alerts": [],
            "signal_strength": "NONE"
        }

    def get_top_movers(self, analyses: List[Dict], limit: int = 10) -> List[Dict]:
        """Get markets with highest velocity"""
        try:
            sorted_analyses = sorted(
                analyses,
                key=lambda x: abs(x.get("velocities", {}).get("1m", 0)),
                reverse=True
            )
            return sorted_analyses[:limit]
        except Exception as e:
            self.logger.error(f"Error getting top movers: {e}")
            return []

    def get_top_signals(self, analyses: List[Dict], limit: int = 10) -> List[Dict]:
        """Get markets with highest signal scores"""
        try:
            sorted_analyses = sorted(
                analyses,
                key=lambda x: x.get("signal_score", 0),
                reverse=True
            )
            return sorted_analyses[:limit]
        except Exception as e:
            self.logger.error(f"Error getting top signals: {e}")
            return []
```

```python
%%writefile modules/chart_builder.py
"""
Chart building functions using Plotly
"""

import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import pandas as pd
import logging
from typing import Dict, List, Optional
import config


class ChartBuilder:
    """Build Plotly charts for the dashboard"""

    def __init__(self):
        self.logger = logging.getLogger(__name__)
        self.template = config.CHART_TEMPLATE

    def create_probability_chart(self, snapshots: List[Dict], title: str = "Probability Over Time") -> go.Figure:
        """Create probability timeline chart"""
        if not snapshots:
            return self._empty_chart("No data available")

        try:
            timestamps = [s['timestamp'] for s in snapshots]
            probabilities = [s['data'].get('probability', 0) for s in snapshots]

            # Convert timestamps to datetime
            from datetime import datetime
            dates = [datetime.fromtimestamp(ts) for ts in timestamps]

            fig = go.Figure()

            fig.add_trace(go.Scatter(
                x=dates,
                y=probabilities,
                mode='lines+markers',
                name='Probability',
                line=dict(color=config.COLORS["accent"], width=2),
                marker=dict(size=4)
            ))

            fig.update_layout(
                title=title,
                xaxis_title="Time",
                yaxis_title="Probability (%)",
                template=self.template,
                height=config.CHART_HEIGHT,
                hovermode='x unified',
                yaxis=dict(range=[0, 100])
            )

            return fig

        except Exception as e:
            self.logger.error(f"Error creating probability chart: {e}")
            return self._empty_chart("Error loading chart")

    def create_velocity_chart(self, analysis: Dict) -> go.Figure:
        """Create velocity comparison chart"""
        try:
            velocities = analysis.get("velocities", {})

            if not velocities:
                return self._empty_chart("No velocity data")

            windows = list(velocities.keys())
            values = [abs(velocities[w]) for w in windows]
            colors = [config.COLORS["positive"] if velocities[w] > 0 else config.COLORS["negative"] for w in windows]

            fig = go.Figure()

            fig.add_trace(go.Bar(
                x=windows,
                y=values,
                marker_color=colors,
                text=[f"{v:.2f}%" for v in values],
                textposition='auto',
            ))

            fig.update_layout(
                title="Velocity Across Time Windows",
                xaxis_title="Time Window",
                yaxis_title="Velocity (% per minute)",
                template=self.template,
                height=300,
            )

            return fig

        except Exception as e:
            self.logger.error(f"Error creating velocity chart: {e}")
            return self._empty_chart("Error loading chart")

    def create_signal_gauge(self, signal_score: float) -> go.Figure:
        """Create signal strength gauge chart"""
        try:
            fig = go.Figure(go.Indicator(
                mode="gauge+number",
                value=signal_score,
                domain={'x': [0, 1], 'y': [0, 1]},
                title={'text': "Signal Strength"},
                gauge={
                    'axis': {'range': [None, 100]},
                    'bar': {'color': self._get_gauge_color(signal_score)},
                    'steps': [
                        {'range': [0, 20], 'color': "#1e2127"},
                        {'range': [20, 40], 'color': "#2a2e37"},
                        {'range': [40, 60], 'color': "#363a47"},
                        {'range': [60, 80], 'color': "#424657"},
                        {'range': [80, 100], 'color': "#4e5267"}
                    ],
                    'threshold': {
                        'line': {'color': "red", 'width': 4},
                        'thickness': 0.75,
                        'value': 80
                    }
                }
            ))

            fig.update_layout(
                template=self.template,
                height=250,
                margin=dict(l=20, r=20, t=40, b=20)
            )

            return fig

        except Exception as e:
            self.logger.error(f"Error creating signal gauge: {e}")
            return self._empty_chart("Error loading gauge")

    def create_volatility_chart(self, snapshots: List[Dict]) -> go.Figure:
        """Create rolling volatility chart"""
        if not snapshots or len(snapshots) < 2:
            return self._empty_chart("Insufficient data")

        try:
            from datetime import datetime

            timestamps = [datetime.fromtimestamp(s['timestamp']) for s in snapshots]
            probabilities = [s['data'].get('probability', 0) for s in snapshots]

            # Calculate rolling volatility
            df = pd.DataFrame({'timestamp': timestamps, 'probability': probabilities})
            df['volatility'] = df['probability'].rolling(window=min(10, len(df)), min_periods=2).std()

            fig = go.Figure()

            fig.add_trace(go.Scatter(
                x=df['timestamp'],
                y=df['volatility'],
                mode='lines',
                name='Volatility',
                fill='tozeroy',
                line=dict(color=config.COLORS["accent"])
            ))

            fig.update_layout(
                title="Probability Volatility",
                xaxis_title="Time",
                yaxis_title="Volatility (StdDev)",
                template=self.template,
                height=300,
                hovermode='x unified'
            )

            return fig

        except Exception as e:
            self.logger.error(f"Error creating volatility chart: {e}")
            return self._empty_chart("Error loading chart")

    def create_volume_chart(self, markets_df: pd.DataFrame, top_n: int = 10) -> go.Figure:
        """Create top markets by volume chart"""
        if markets_df.empty:
            return self._empty_chart("No market data")

        try:
            top_markets = markets_df.nlargest(top_n, 'volume')

            fig = go.Figure()

            fig.add_trace(go.Bar(
                x=top_markets['volume'],
                y=top_markets['question'].apply(lambda x: x[:50] + '...' if len(x) > 50 else x),
                orientation='h',
                marker_color=config.COLORS["positive"]
            ))

            fig.update_layout(
                title=f"Top {top_n} Markets by Volume",
                xaxis_title="Volume ($)",
                yaxis_title="Market",
                template=self.template,
                height=400,
                margin=dict(l=200)
            )

            return fig

        except Exception as e:
            self.logger.error(f"Error creating volume chart: {e}")
            return self._empty_chart("Error loading chart")

    def _get_gauge_color(self, score: float) -> str:
        """Get color for gauge based on score"""
        if score >= 80:
            return "#ff4444"
        elif score >= 60:
            return "#ffaa00"
        elif score >= 40:
            return "#ffff00"
        else:
            return "#888888"

    def _empty_chart(self, message: str) -> go.Figure:
        """Create empty chart with message"""
        fig = go.Figure()

        fig.add_annotation(
            text=message,
            xref="paper",
            yref="paper",
            x=0.5,
            y=0.5,
            showarrow=False,
            font=dict(size=16, color=config.COLORS["neutral"])
        )

        fig.update_layout(
            template=self.template,
            height=config.CHART_HEIGHT,
            xaxis=dict(visible=False),
            yaxis=dict(visible=False)
        )

        return fig
```

```python
%%writefile app.py
"""
Polymarket Analytics Dashboard - Main Application
Advanced movement detection and market analysis
"""

import streamlit as st
import pandas as pd
import time
from datetime import datetime, timedelta
import logging

# Import modules
import config
from modules import (
    PolymarketAPI,
    DataProcessor,
    SignalEngine,
    StorageManager,
    ChartBuilder,
    setup_logging,
    format_number,
    format_percentage
)
from modules.utils import get_color_for_change

# Setup logging
logger = setup_logging()

# Page configuration
st.set_page_config(
    page_title="Polymarket Analytics",
    page_icon="📊",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Custom CSS for dark mode
st.markdown(f"""
    <style>
        .stApp {{
            background-color: {config.COLORS['background']};
        }}
        .metric-card {{
            background-color: {config.COLORS['secondary_bg']};
            padding: 20px;
            border-radius: 10px;
            margin: 10px 0;
        }}
        .signal-strong {{
            color: {config.COLORS['negative']};
            font-weight: bold;
            font-size: 1.2em;
        }}
        .signal-moderate {{
            color: {config.COLORS['neutral']};
            font-weight: bold;
        }}
        .positive {{
            color: {config.COLORS['positive']};
        }}
        .negative {{
            color: {config.COLORS['negative']};
        }}
        .alert-box {{
            background-color: {config.COLORS['secondary_bg']};
            border-left: 4px solid {config.COLORS['negative']};
            padding: 15px;
            margin: 10px 0;
            border-radius: 5px;
        }}
    </style>
""", unsafe_allow_html=True)


# Initialize session state
if 'api_client' not in st.session_state:
    st.session_state.api_client = PolymarketAPI()
    st.session_state.storage = StorageManager()
    st.session_state.processor = DataProcessor()
    st.session_state.signal_engine = SignalEngine(st.session_state.processor)
    st.session_state.chart_builder = ChartBuilder()
    st.session_state.last_refresh = None
    st.session_state.markets_df = pd.DataFrame()
    st.session_state.analyses = []


def fetch_and_process_markets():
    """Fetch markets and process data"""
    try:
        with st.spinner("Fetching market data..."):
            # Fetch markets
            markets_raw = st.session_state.api_client.get_markets(limit=50)

            if not markets_raw:
                st.error("Failed to fetch markets. Please check API connection.")
                return False

            # Process markets
            st.session_state.markets_df = st.session_state.processor.process_markets(markets_raw)

            if st.session_state.markets_df.empty:
                st.warning("No markets to display")
                return False

            # Save snapshots and analyze
            st.session_state.analyses = []

            for _, market in st.session_state.markets_df.iterrows():
                market_dict = market.to_dict()

                # Save snapshot
                st.session_state.storage.save_snapshot(
                    market['market_id'],
                    market_dict
                )

                # Get historical snapshots
                snapshots = st.session_state.storage.get_snapshots(
                    market['market_id'],
                    limit=100
                )

                # Analyze market
                if len(snapshots) >= 2:
                    analysis = st.session_state.signal_engine.analyze_market(
                        market_dict,
                        snapshots
                    )
                    st.session_state.analyses.append(analysis)

            st.session_state.last_refresh = datetime.now()
            return True

    except Exception as e:
        logger.error(f"Error fetching markets: {e}")
        st.error(f"Error: {e}")
        return False


def render_top_metrics():
    """Render top-level metrics"""
    col1, col2, col3, col4 = st.columns(4)

    with col1:
        st.metric(
            "Total Markets",
            len(st.session_state.markets_df),
            delta=None
        )

    with col2:
        total_volume = st.session_state.markets_df['volume'].sum()
        st.metric(
            "Total Volume",
            format_number(total_volume),
            delta=None
        )

    with col3:
        if st.session_state.analyses:
            strong_signals = len([a for a in st.session_state.analyses if a['signal_score'] >= 60])
            st.metric(
                "Strong Signals",
                strong_signals,
                delta=None
            )
        else:
            st.metric("Strong Signals", "0", delta=None)

    with col4:
        if st.session_state.last_refresh:
            time_ago = (datetime.now() - st.session_state.last_refresh).seconds
            st.metric(
                "Last Updated",
                f"{time_ago}s ago",
                delta=None
            )
        else:
            st.metric("Last Updated", "Never", delta=None)


def render_signals_section():
    """Render top signals section"""
    st.header("🎯 Top Signals - Rapid Repricing Detection")

    if not st.session_state.analyses:
        st.info("Run initial data fetch to see signals")
        return

    top_signals = st.session_state.signal_engine.get_top_signals(
        st.session_state.analyses,
        limit=10
    )

    for analysis in top_signals:
        if analysis['signal_score'] < 20:
            continue

        with st.expander(
            f"📊 {analysis['question'][:80]}... - Score: {analysis['signal_score']:.1f}",
            expanded=analysis['signal_score'] >= 60
        ):
            col1, col2 = st.columns([2, 1])

            with col1:
                # Alerts
                if analysis['alerts']:
                    st.markdown("**🚨 Alerts:**")
                    for alert in analysis['alerts']:
                        st.markdown(f"- {alert}")
                    st.markdown("---")

                # Metrics
                metrics_col1, metrics_col2, metrics_col3 = st.columns(3)

                with metrics_col1:
                    st.metric(
                        "Current Probability",
                        f"{analysis['current_probability']:.1f}%"
                    )

                with metrics_col2:
                    velocity_1m = analysis['velocities'].get('1m', 0)
                    st.metric(
                        "1m Velocity",
                        f"{velocity_1m:+.2f}%/min",
                        delta=None
                    )

                with metrics_col3:
                    st.metric(
                        "Signal Strength",
                        analysis['signal_strength']
                    )

                # Velocity breakdown
                st.markdown("**Velocity Across Time Windows:**")
                vel_cols = st.columns(4)
                for i, (window, vel) in enumerate(analysis['velocities'].items()):
                    with vel_cols[i]:
                        st.metric(window, f"{vel:+.2f}%/min")

                # Additional metrics
                st.markdown("**Analysis Metrics:**")
                metric_cols = st.columns(4)
                with metric_cols[0]:
                    st.metric("Acceleration", f"{analysis['acceleration']:+.2f}")
                with metric_cols[1]:
                    st.metric("Volatility", f"{analysis['volatility']:.2f}%")
                with metric_cols[2]:
                    st.metric("Consistency", f"{analysis['consistency']*100:.0f}%")
                with metric_cols[3]:
                    consolidating = analysis['consolidation'].get('is_consolidating', False)
                    st.metric("Consolidating", "Yes" if consolidating else "No")

            with col2:
                # Signal gauge
                gauge = st.session_state.chart_builder.create_signal_gauge(
                    analysis['signal_score']
                )
                st.plotly_chart(gauge, use_container_width=True)


def render_movers_section():
    """Render biggest movers section"""
    st.header("🚀 Biggest Movers")

    if not st.session_state.analyses:
        st.info("Run initial data fetch to see movers")
        return

    top_movers = st.session_state.signal_engine.get_top_movers(
        st.session_state.analyses,
        limit=10
    )

    # Create table
    movers_data = []
    for analysis in top_movers:
        velocity_1m = analysis['velocities'].get('1m', 0)

        if abs(velocity_1m) < 0.5:
            continue

        movers_data.append({
            'Market': analysis['question'][:60] + '...',
            'Probability': f"{analysis['current_probability']:.1f}%",
            '1m Velocity': f"{velocity_1m:+.2f}%/min",
            'Signal Score': f"{analysis['signal_score']:.1f}",
            'Alerts': len(analysis['alerts'])
        })

    if movers_data:
        df_movers = pd.DataFrame(movers_data)
        st.dataframe(df_movers, use_container_width=True, hide_index=True)
    else:
        st.info("No significant movements detected")


def render_markets_table():
    """Render main markets table"""
    st.header("📈 All Markets")

    if st.session_state.markets_df.empty:
        st.info("No markets loaded yet")
        return

    # Filters
    col1, col2, col3 = st.columns(3)

    with col1:
        search_query = st.text_input("🔍 Search markets", "")

    with col2:
        sort_by = st.selectbox(
            "Sort by",
            ["Volume", "Liquidity", "Probability", "Signal Score"]
        )

    with col3:
        min_volume = st.number_input("Min Volume ($)", min_value=0, value=0, step=1000)

    # Filter dataframe
    df_filtered = st.session_state.markets_df.copy()

    if search_query:
        df_filtered = df_filtered[
            df_filtered['question'].str.contains(search_query, case=False, na=False)
        ]

    if min_volume > 0:
        df_filtered = df_filtered[df_filtered['volume'] >= min_volume]

    # Add signal scores
    if st.session_state.analyses:
        signal_scores = {a['market_id']: a['signal_score'] for a in st.session_state.analyses}
        df_filtered['signal_score'] = df_filtered['market_id'].map(signal_scores).fillna(0)
    else:
        df_filtered['signal_score'] = 0

    # Sort
    sort_columns = {
        "Volume": "volume",
        "Liquidity": "liquidity",
        "Probability": "probability",
        "Signal Score": "signal_score"
    }

    df_filtered = df_filtered.sort_values(sort_columns[sort_by], ascending=False)

    # Display table
    display_df = df_filtered[[
        'question', 'probability', 'yes_price', 'no_price',
        'volume_formatted', 'liquidity_formatted', 'signal_score'
    ]].copy()

    display_df.columns = [
        'Market', 'Probability (%)', 'YES Price', 'NO Price',
        'Volume', 'Liquidity', 'Signal Score'
    ]

    st.dataframe(
        display_df,
        use_container_width=True,
        hide_index=True,
        height=400
    )


def render_market_detail(market_id: str):
    """Render detailed view for a specific market"""
    market = st.session_state.markets_df[
        st.session_state.markets_df['market_id'] == market_id
    ].iloc[0]

    st.header(f"📊 {market['question']}")

    # Get snapshots and analysis
    snapshots = st.session_state.storage.get_snapshots(market_id, limit=100)

    if len(snapshots) < 2:
        st.warning("Not enough historical data for this market")
        return

    analysis = next(
        (a for a in st.session_state.analyses if a['market_id'] == market_id),
        None
    )

    if not analysis:
        st.warning("Analysis not available for this market")
        return

    # Metrics row
    col1, col2, col3, col4 = st.columns(4)

    with col1:
        st.metric("Probability", f"{market['probability']:.1f}%")

    with col2:
        st.metric("Volume", format_number(market['volume']))

    with col3:
        st.metric("Liquidity", format_number(market['liquidity']))

    with col4:
        st.metric("Signal Score", f"{analysis['signal_score']:.1f}")

    # Charts
    tab1, tab2, tab3, tab4 = st.tabs([
        "Probability Timeline",
        "Velocity Analysis",
        "Volatility",
        "Raw Data"
    ])

    with tab1:
        prob_chart = st.session_state.chart_builder.create_probability_chart(
            snapshots,
            f"Probability Over Time - {market['question'][:50]}"
        )
        st.plotly_chart(prob_chart, use_container_width=True)

    with tab2:
        vel_chart = st.session_state.chart_builder.create_velocity_chart(analysis)
        st.plotly_chart(vel_chart, use_container_width=True)

    with tab3:
        vol_chart = st.session_state.chart_builder.create_volatility_chart(snapshots)
        st.plotly_chart(vol_chart, use_container_width=True)

    with tab4:
        st.json(analysis)


def render_sidebar():
    """Render sidebar controls"""
    with st.sidebar:
        st.title("⚙️ Controls")

        # Refresh button
        if st.button("🔄 Refresh Data", use_container_width=True):
            fetch_and_process_markets()
            st.rerun()

        # Auto-refresh toggle
        auto_refresh = st.checkbox("Auto-refresh (60s)", value=False)

        if auto_refresh:
            time.sleep(60)
            fetch_and_process_markets()
            st.rerun()

        st.markdown("---")

        # Stats
        st.subheader("📊 Stats")

        if st.session_state.last_refresh:
            st.text(f"Last refresh: {st.session_state.last_refresh.strftime('%H:%M:%S')}")

        storage_stats = st.session_state.storage.get_storage_stats()
        if storage_stats:
            st.text(f"Markets tracked: {storage_stats.get('markets_tracked', 0)}")
            st.text(f"Total snapshots: {storage_stats.get('total_snapshots', 0)}")
            st.text(f"Storage: {storage_stats.get('storage_mb', 0):.2f} MB")

        cache_stats = st.session_state.api_client.get_cache_stats()
        st.text(f"Cached requests: {cache_stats.get('cached_requests', 0)}")

        st.markdown("---")

        # Actions
        st.subheader("🛠️ Actions")

        if st.button("Clear API Cache", use_container_width=True):
            st.session_state.api_client.clear_cache()
            st.success("Cache cleared")

        if st.button("Cleanup Old Data", use_container_width=True):
            st.session_state.storage.cleanup_old_snapshots()
            st.success("Cleanup complete")

        st.markdown("---")

        # Settings
        st.subheader("⚙️ Settings")
        st.text(f"Refresh interval: {config.AUTO_REFRESH_SECONDS}s")
        st.text(f"Max markets: {config.MAX_MARKETS_TO_TRACK}")
        st.text(f"Cache duration: {config.CACHE_DURATION_SECONDS}s")


def main():
    """Main application"""

    # Title
    st.title("📊 Polymarket Analytics Dashboard")
    st.markdown("*Real-time market analysis with rapid repricing detection*")

    # Render sidebar
    render_sidebar()

    # Initial load
    if st.session_state.markets_df.empty:
        if st.button("🚀 Load Markets", use_container_width=True, type="primary"):
            if fetch_and_process_markets():
                st.success("Markets loaded successfully!")
                st.rerun()
        return

    # Top metrics
    render_top_metrics()

    st.markdown("---")

    # Main sections
    render_signals_section()

    st.markdown("---")

    render_movers_section()

    st.markdown("---")

    render_markets_table()

    # Market detail (if selected)
    if st.session_state.markets_df is not None and not st.session_state.markets_df.empty:
        st.markdown("---")
        st.subheader("🔍 Market Details")

        selected_market = st.selectbox(
            "Select market for detailed analysis",
            options=st.session_state.markets_df['market_id'].tolist(),
            format_func=lambda x: st.session_state.markets_df[
                st.session_state.markets_df['market_id'] == x
            ]['question'].iloc[0][:80] + "..."
        )

        if selected_market:
            render_market_detail(selected_market)


if __name__ == "__main__":
    main()
```
